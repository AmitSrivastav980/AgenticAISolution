```python
#%pip install python-dotenv  -- this command is for installing dot env module

from dotenv import load_dotenv
from IPython.display import Markdown, display
import os
```

```python
# Next it's time to load the API keys into environment variables
# If this returns false, see the next cell!
load_dotenv(override=True)
```

```python
gemini_api_key = os.getenv("GEMINI_API_KEY")
gemini_base_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

#%pip install openai
from openai import OpenAI
gemini_client = OpenAI(
    base_url=gemini_base_url,
    api_key=gemini_api_key
)
content = "pick a business area that might be worth exploring for an Agentic AI opportunity"
messages = [{"role": "user", "content": content}]


```

```python

response = gemini_client.chat.completions.create(
    model="gemini-2.5-flash-lite",
    messages=messages
)
message = response.choices[0].message.content
display((Markdown(response.choices[0].message.content)))
```

```python
question = "could you please present a pain-point in that industry something challenging that might be ripe for an Agentic solution)"
messages += [{"role": "user", "content": question}]

response = gemini_client.chat.completions.create(
    model="gemini-2.5-flash-lite",
    messages=messages)
message = response.choices[0].message.content
display((Markdown(message)))
display((Markdown(message)))
```
